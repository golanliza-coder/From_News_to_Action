```python
import pandas as pd
import os
from google.colab import drive

# 1. חיבור בטוח לגוגל דרייב שלך (יפתח חלון אישור קטן)
drive.mount('/content/drive')

# 2. הגדרת נתיב התיקייה המדויק שבו נמצאים קבצי האקסל
# שנה את השם לאחר ה-MyDrive לתיקייה האמיתית שלך בדרייב
DATA_DIR = "/content/drive/MyDrive/From_News_to_Action/DataSets/BaseLine"

# 3. רשימת הקבצים לאיחוד
file_list = [
    't01_01.xlsx', 't01_02.xlsx', 't01_03.xlsx', 't01_04.xlsx',
    't01_05.xlsx', 't01_06.xlsx', 't01_07.xlsx', 't01_08.xlsx',
    't01_09.xlsx', 't01_10.xlsx', 't01_11.xlsx', 't01_12.xlsx',
    't01_13.xlsx', 't01_14.xlsx', 'sep 2026.xlsx'
]

# 4. טעינת הקבצים תוך שימוש בנתיב המלא
all_dfs = []
for file in file_list:
    # בניית הנתיב המלא לקובץ באמצעות os.path.join
    full_path = os.path.join(DATA_DIR, file)

    try:
        df = pd.read_excel(full_path)
        all_dfs.append(df)
        print(f"הקובץ {file} נטען בהצלחה. שורות: {df.shape}")
    except Exception as e:
        print(f"שגיאה בטעינת הקובץ {file} מתוך הנתיב {full_path}: {e}")

# 5. איחוד המאגר
if all_dfs:
    df_accidents = pd.concat(all_dfs, axis=0, ignore_index=True)
    print(f"\nכל הקבצים אוחדו! מבנה המאגר הסופי: {df_accidents.shape}")
else:
    print("\nלא נטענו קבצים. ודא שהנתיב לתיקייה בדרייב מדויק לחלוטין.")
```

```python
import pandas as pd
import os

DATA_DIR = "/content/drive/MyDrive/From_News_to_Action/DataSets/BaseLine"

file_list = [
    't01_01.xlsx', 't01_02.xlsx', 't01_03.xlsx', 't01_04.xlsx',
    't01_05.xlsx', 't01_06.xlsx', 't01_07.xlsx', 't01_08.xlsx',
    't01_09.xlsx', 't01_10.xlsx', 't01_11.xlsx', 't01_12.xlsx',
    't01_13.xlsx', 't01_14.xlsx', 'sep 2026.xlsx'
]

print("🕵️ מתחיל בסריקה ובדיקת העיצוב של 15 הקבצים...")
print("=" * 60)

for file in file_list:
    full_path = os.path.join(DATA_DIR, file)
    try:
        # טעינת 15 שורות ראשונות ללא הגדרת כותרת כדי לחקור את המבנה הגולמי
        df_raw = pd.read_excel(full_path, header=None, nrows=15)

        # זיהוי שורות מעניינות: שורות שאינן ריקות לחלוטין
        non_empty_rows = df_raw.dropna(how='all').index.tolist()

        print(f"📋 קובץ: {file}")
        print(f"   • מימדים גולמיים (Rows, Cols): {df_raw.shape}")
        print(f"   • שורות המכילות טקסט (אינדקס): {non_empty_rows[:6]}")

        # הצגת 3 השורות הראשונות שיש בהן נתונים כדי שנוכל לראות את הכותרת בעיניים
        print("   • מבנה שורות עליונות:")
        for idx in non_empty_rows[:3]:
            row_content = df_raw.iloc[idx].dropna().tolist()[:4] # 4 ערכים ראשונים בשורה
            print(f"     [שורה {idx}]: {row_content}")

        print("-" * 60)

    except Exception as e:
        print(f"❌ כשל בסריקת הקובץ {file}: {e}")
        print("-" * 60)
```

```python
import pandas as pd
import os

DATA_DIR = "/content/drive/MyDrive/From_News_to_Action/DataSets/BaseLine"

# מיפוי ידני ומדויק של מספר השורות שיש לדלג עליהן בכל קובץ (על סמך ניסוי הסריקה שלנו)
# המטרה: להגיע ישירות לשורת הנתונים הראשונה
skip_rows_mapping = {
    't01_01': 3, 't01_02': 3, 't01_03': 3, 't01_04': 3, 't01_05': 3,
    't01_06': 3, 't01_07': 3, 't01_08': 3, 't01_09': 3, 't01_10': 3,
    't01_11': 3, 't01_12': 3, 't01_13': 3, 't01_14': 4, 'sep 2026': 5
}

validated_datasets = {}

print("⚙️ מריץ טעינה סופית ומאובטחת לשמירה על כל עמודות המידע...")
print("=" * 60)

for file_prefix, skip_count in skip_rows_mapping.items():
    file = f"{file_prefix}.xlsx"
    full_path = os.path.join(DATA_DIR, file)

    try:
        # טעינה ללא הגדרת כותרת כדי למנוע מחיקת עמודות Unnamed
        df_clean = pd.read_excel(full_path, header=None, skiprows=skip_count)

        # הסרת שורות ועמודות שכולן ריקות לחלוטין (NaN)
        df_clean = df_clean.dropna(how='all').dropna(axis=1, how='all')

        # אינדוקס העמודות בצורה נקייה: עמודה 0 תהיה המשתנה המילולי, ושאר העמודות הן הנתונים
        df_clean.columns = [f"col_{i}" for i in range(df_clean.shape[1])]

        validated_datasets[file_prefix] = df_clean
        print(f"✅ {file} נטען בהצלחה! מימדים סופיים: {df_clean.shape}")

    except Exception as e:
        print(f"❌ כשל בטעינת {file}: {e}")

print(f"\n🎉 המאגרים מוכנים ומלאים בתוך validated_datasets! אפשר להמשיך הלאה.")
```
