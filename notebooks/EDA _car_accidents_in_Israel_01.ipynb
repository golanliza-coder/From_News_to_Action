```python
import psycopg2
import pandas as pd
import numpy as np
import json
from google.colab import userdata

DATABASE_URL = userdata.get('DATABASE_URL')
conn = psycopg2.connect(DATABASE_URL)

df = pd.read_sql("SELECT id, published_date, article_data FROM traffic_incidents;", conn)
conn.close()

def extract_coords(val):
    if isinstance(val, str):
        val = json.loads(val)
    if not isinstance(val, dict):
        return None, None

    # חיפוש ברמה הראשית של ה-JSON
    lat = val.get('lat') or val.get('latitude')
    lng = val.get('lng') or val.get('lon') or val.get('longitude')

    # חיפוש בתוך תת-מילון location אם קיים
    if lat is None and isinstance(val.get('location'), dict):
        loc = val['location']
        lat = loc.get('lat') or loc.get('latitude')
        lng = loc.get('lng') or loc.get('lon') or loc.get('longitude')

    return lat, lng

coords = df['article_data'].apply(extract_coords)
df['lat'] = pd.to_numeric([c[0] for c in coords], errors='coerce')
df['lng'] = pd.to_numeric([c[1] for c in coords], errors='coerce')

df_valid = df.dropna(subset=['lat', 'lng']).reset_index(drop=True)
df_valid['published_date'] = pd.to_datetime(df_valid['published_date'], errors='coerce')

# חישוב מרחקים במטרים (נוסחת Haversine)
lats = np.radians(df_valid['lat'].values)
lons = np.radians(df_valid['lng'].values)

dlat = lats[:, None] - lats[None, :]
dlon = lons[:, None] - lons[None, :]
a = np.sin(dlat / 2)**2 + np.cos(lats[:, None]) * np.cos(lats[None, :]) * np.sin(dlon / 2)**2
dist_meters = 6371000 * 2 * np.arcsin(np.sqrt(a))

# תנאי: מרחק עד 500 מטר (ולא אותה שורה עצמה)
within_500m = (dist_meters <= 500) & (np.eye(len(df_valid)) == 0)

# סינון לפי חלון זמן של 90 יום
if df_valid['published_date'].notna().any():
    dates = df_valid['published_date'].values
    time_diff_days = np.abs((dates[:, None] - dates[None, :]) / np.timedelta64(1, 'D'))
    within_90d = time_diff_days <= 90
    valid_neighbors = within_500m & within_90d
else:
    valid_neighbors = within_500m

neighbor_counts = valid_neighbors.sum(axis=1)

total = len(df_valid)
orange_or_red = int((neighbor_counts >= 1).sum()) # 2 תאונות ומעלה במקבץ
red_only = int((neighbor_counts >= 2).sum())      # 3 תאונות ומעלה במקבץ (מפגע תשתית חשוד)

print("-" * 50)
print(f"סה״כ אירועים מאומתים במפה: {total}")
print(f"אירועים במוקד חוזר (כתום + אדום, 2+ תאונות ברדיוס 500 מ'): {orange_or_red} ({orange_or_red/total*100:.1f}%)")
print(f"אירועים במוקד אדום (מפגע תשתית חשוד, 3+ תאונות ברדיוס 500 מ'): {red_only} ({red_only/total*100:.1f}%)")
```

```python
import psycopg2
import pandas as pd
import numpy as np
import json
from google.colab import userdata

DATABASE_URL = userdata.get('DATABASE_URL')
conn = psycopg2.connect(DATABASE_URL)

df = pd.read_sql("SELECT id, published_date, source, article_data FROM traffic_incidents;", conn)
conn.close()

def parse_record(val):
    if isinstance(val, str):
        val = json.loads(val)
    if not isinstance(val, dict):
        return None, None, "", ""

    lat = val.get('lat') or val.get('latitude')
    lng = val.get('lng') or val.get('lon') or val.get('longitude')
    loc_name = val.get('verified_address') or val.get('resolved_location') or val.get('location_raw_text') or ""
    raw_text = val.get('raw_text') or val.get('summary') or ""

    if lat is None and isinstance(val.get('location'), dict):
        loc = val['location']
        lat = loc.get('lat') or loc.get('latitude')
        lng = loc.get('lng') or loc.get('lon') or loc.get('longitude')
        loc_name = loc_name or loc.get('formatted_address') or loc.get('text_description') or ""

    return lat, lng, str(loc_name), str(raw_text)

parsed = df['article_data'].apply(parse_record)
df['lat'] = pd.to_numeric([p[0] for p in parsed], errors='coerce')
df['lng'] = pd.to_numeric([p[1] for p in parsed], errors='coerce')
df['loc_name'] = [p[2] for p in parsed]
df['raw_text'] = [p[3] for p in parsed]

df_valid = df.dropna(subset=['lat', 'lng']).reset_index(drop=True)
df_valid['published_date'] = pd.to_datetime(df_valid['published_date'], errors='coerce')

# פונקציה לזיהוי כתובות גנריות ברמת עיר בלבד (למשל "Haifa, Israel" ללא פסיק נוסף של רחוב/כביש)
def is_generic_city(addr):
    addr_clean = addr.strip()
    # אם הכתובת באנגלית בפורמט "City, Israel" (פסיק אחד בלבד) ואין בה מספר כביש או צומת
    if addr_clean.endswith(", Israel") and addr_clean.count(",") == 1:
        if not any(k in addr_clean.lower() for k in ['road', 'hwy', 'highway', 'route', 'st', 'blvd', 'interchange', 'junction', 'כביש', 'מחלף', 'צומת', 'רחוב', 'שדרות']):
            return True
    return False

df_valid['is_generic'] = df_valid['loc_name'].apply(is_generic_city)
df_specific = df_valid[~df_valid['is_generic']].reset_index(drop=True)

print(f"סה״כ אירועים עם נ״צ: {len(df_valid)}")
print(f"אירועים שסוננו כמרכז-עיר כללי בלבד: {df_valid['is_generic'].sum()}")
print(f"אירועים במיקום ממוקד (רחוב / צומת / כביש): {len(df_specific)}")
print("=" * 60)

# חישוב מרחקים במטרים על המיקומים הממוקדים בלבד
lats = np.radians(df_specific['lat'].values)
lons = np.radians(df_specific['lng'].values)

dlat = lats[:, None] - lats[None, :]
dlon = lons[:, None] - lons[None, :]
a = np.sin(dlat / 2)**2 + np.cos(lats[:, None]) * np.cos(lats[None, :]) * np.sin(dlon / 2)**2
dist_meters = 6371000 * 2 * np.arcsin(np.sqrt(a))

within_500m = (dist_meters <= 500) & (np.eye(len(df_specific)) == 0)

if df_specific['published_date'].notna().any():
    dates = df_specific['published_date'].values
    time_diff_days = np.abs((dates[:, None] - dates[None, :]) / np.timedelta64(1, 'D'))
    valid_neighbors = within_500m & (time_diff_days <= 90)
else:
    valid_neighbors = within_500m

df_specific['cluster_size'] = valid_neighbors.sum(axis=1) + 1

# חישוב אחוזים נקיים מהטיה
red_incidents = df_specific[df_specific['cluster_size'] >= 3]
print(f"שיעור התאונות במוקדים אדומים (לאחר ניקוי מרכזי ערים): {len(red_incidents)} מתוך {len(df_specific)} ({len(red_incidents)/len(df_specific)*100:.1f}%)")
print("=" * 60)

# איתור 3 המוקדים האדומים המובילים (ללא חפיפה בין מקבצים)
visited = set()
top_hotspots = []

# מיון מהמקבץ הגדול ביותר לקטן ביותר
sorted_indices = df_specific.sort_values(by='cluster_size', ascending=False).index

for idx in sorted_indices:
    if idx in visited or df_specific.loc[idx, 'cluster_size'] < 3:
        continue
    # מציאת כל השכנים במקבץ הזה
    neighbors = np.where(valid_neighbors[idx])[0].tolist()
    cluster_members = [idx] + neighbors
    visited.update(cluster_members)

    top_hotspots.append({
        'location': df_specific.loc[idx, 'loc_name'],
        'count': len(cluster_members),
        'lat': df_specific.loc[idx, 'lat'],
        'lng': df_specific.loc[idx, 'lng'],
        'sample_texts': df_specific.loc[cluster_members[:3], 'raw_text'].tolist()
    })
    if len(top_hotspots) == 3:
        break

print("🔴 3 המוקדים האדומים המובילים (מפגעי תשתית חשודים):")
for i, spot in enumerate(top_hotspots, 1):
    print(f"\nמקום {i}: {spot['location']} — {spot['count']} תאונות ברדיוס 500 מ'!")
    print(f"   נ״צ: ({spot['lat']:.4f}, {spot['lng']:.4f})")
    for j, txt in enumerate(spot['sample_texts'], 1):
        clean_txt = txt.replace('\n', ' ')[:110]
        print(f"   • אירוע {j}: {clean_txt}...")
```

```python
import psycopg2
import pandas as pd
import json
from google.colab import userdata

DATABASE_URL = userdata.get('DATABASE_URL')
conn = psycopg2.connect(DATABASE_URL)

df_raw = pd.read_sql("SELECT id, source, published_date, article_data FROM traffic_incidents;", conn)
conn.close()

exclude_keywords = (
    'הפגנה', 'מחאה', 'תקיפה', 'עזה', 'פיגוע', 'קטטה',
    'ירי', 'רקטה', 'טיל', 'דקירה', 'שריפה בדירה', 'טבע למוות'
)

filtered_out = []
for _, r in df_raw.iterrows():
    d = json.loads(r['article_data']) if isinstance(r['article_data'], str) else r['article_data']
    raw_txt = str(d.get('raw_text', ''))

    matched_words = [w for w in exclude_keywords if w in raw_txt]
    if d.get('is_traffic_accident') is False or matched_words:
        filtered_out.append({
            'id': r['id'],
            'source': r['source'],
            'matched_word': ", ".join(matched_words) if matched_words else "LLM False",
            'text_preview': raw_txt.replace('\n', ' ')[:120]
        })

df_filtered = pd.DataFrame(filtered_out).drop_duplicates(subset=['id'])
print(f"סה״כ רשומות שסוננו החוצה: {len(df_filtered)}")
print("=" * 80)
for idx, row in df_filtered.iterrows():
    print(f"[{row['source']}] מילה שזוהתה: ({row['matched_word']}) -> {row['text_preview']}...")
```
